```python
from cassandra.cluster import Cluster
import warnings

warnings.filterwarnings('ignore')

try:
    print("Tentativo di connessione a Cassandra...")
    # Aggiunto 'localhost' e un timeout per evitare caricamenti infiniti
    cluster = Cluster(['127.0.0.1', 'localhost'], port=9042, connect_timeout=10)
    session = cluster.connect()
    
    print("✅ SUCCESSO! Connesso a Cassandra.")
    
    row = session.execute("SELECT cluster_name, data_center FROM system.local").one()
    print(f"Cluster Name: {row.cluster_name}")
    print(f"Data Center: {row.data_center}")
    
except Exception as e:
    print(f"❌ FALLITO. Errore di connessione: {e}")
finally:
    if 'cluster' in locals():
        cluster.shutdown()
        print("Connessione chiusa.")
```
