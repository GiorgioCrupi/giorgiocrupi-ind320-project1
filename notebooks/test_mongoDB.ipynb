```python
import os
import datetime
from dotenv import load_dotenv
from pymongo import MongoClient

# Load the MongoDB connection string from the hidden .env file
load_dotenv()
MONGO_URI = os.getenv("MONGO_URI")

try:
    print("Connecting to MongoDB Atlas...")
    client = MongoClient(MONGO_URI)

    db = client['ind320_test_db']
    collection = db['test_collection']

    # Insert a test document
    test_document = {
        "author": "Giorgio Crupi",
        "project": "IND320 Assignment 2",
        "test_date": datetime.datetime.now(),
        "status": "Success"
    }
    insert_result = collection.insert_one(test_document)
    print(f"✅ Document inserted with ID: {insert_result.inserted_id}")

    # Read the document back from the cloud
    retrieved_data = collection.find_one({"author": "Giorgio Crupi"})
    print(f"✅ Document read from the cloud: {retrieved_data['project']} (test date: {retrieved_data['test_date']})")

    # Remove the test document to keep the database clean
    collection.delete_one({"_id": insert_result.inserted_id})
    print("🧹 Test document deleted.")

except Exception as e:
    print(f"❌ MongoDB error: {e}")
```
